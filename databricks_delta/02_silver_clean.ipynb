```python
from pyspark.sql.types import *
```

```python
schema = StructType([
  StructField("request_id", StringType()),
  StructField("event_time", StringType()),
  StructField("service", StringType()),
  StructField("endpoint", StringType()),
  StructField("method", StringType()),
  StructField("status_code", IntegerType()),
  StructField("latency_ms", IntegerType()),
  StructField("bytes_in", IntegerType()),
  StructField("bytes_out", IntegerType()),
  StructField("client_type", StringType()),
  StructField("region", StringType()),
  StructField("host", StringType())
  ])
```

```python
from pyspark.sql.functions import from_json, col
```

```python
parsed = bronze.withColumn("data", from_json(col("value"),schema))
```

```python
display(parsed.select("data"))
```

```python
clean = parsed.select("data.*")
```

```python
display(clean)
```

# step 2

```python
good = parsed.filter(
    col("data").isNotNull() &
    col("data.request_id").isNotNull() &
    (col("data.latency_ms") >= 0) &
    col("data.status_code").between(100, 599)
)

bad = parsed.subtract(good)
```

```python
clean = good.select("data.*")
display(clean)
```

```python
from pyspark.sql.window import Window
```

```python
# Step 1: define how to group duplicates
window_spec = Window.partitionBy("request_id").orderBy(col("event_time").desc())
```

```python
from pyspark.sql.functions import row_number
```

```python
# Step 2: rank rows within each request_id
deduped = clean.withColumn("row_rank",row_number().over(window_spec))
```

```python
# Step 3: keep only the latest record per request_id
deduped = deduped.filter(col("row_rank")==1).drop("row_rank")
display(deduped)
```

```python
deduped.write.format("delta").mode("overwrite").saveAsTable("workspace.api_logs_schema.silver_api_logs_clean")
```
