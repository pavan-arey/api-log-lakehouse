```python
df = spark.read.text("/Volumes/workspace/api_logs_schema/api_logs_volume/")
```

```python
display(df.head(5))
```

```python
display(df.count())
```

```python
from pyspark.sql.functions import current_timestamp, col
```

```python
bronze_df = df.withColumn("ingest_ts", current_timestamp()).withColumn("source_file",col("_metadata.file_path"))
```

```python
display(bronze_df.head(1))
```

```python
bronze_df.write.format("delta").mode("overwrite").saveAsTable("workspace.api_logs_schema.bronze_api_logs_raw")
```
